```python
PYTHAGOREAN_ALPHABET = {
    "A": 1, "B": 2, "C": 3, "D": 4, "E": 5, "F": 6, "G": 7, "H": 8, "I": 9,
    "J": 1, "K": 2, "L": 3, "M": 4, "N": 5, "O": 6, "P": 7, "Q": 8, "R": 9,
    "S": 1, "T": 2, "U": 3, "V": 4, "W": 5, "X": 6, "Y": 7, "Z": 8,
}
```

```python
len(PYTHAGOREAN_ALPHABET)
```

```python
sorted(PYTHAGOREAN_ALPHABET.keys())
```

```python
MASTER_NUMBERS = {11, 22, 33}
```

```python
def reduce_number(number):
    while number > 9 and number not in MASTER_NUMBERS:
        number = sum(int(digit) for digit in str(number))
    return number
```

```python
reduce_number(38)
```

```python
reduce_number(49)
```

```python
def calculate_letters_value(word):
    total = 0

    for letter in word.upper():
        if letter in PYTHAGOREAN_ALPHABET:
            total += PYTHAGOREAN_ALPHABET[letter]

    return total
```

```python
calculate_letters_value("BENJAMIN")
```

```python
reduce_number(calculate_letters_value("BENJAMIN"))
```
